# 03 – Posterior Maps

Fine-grid posterior mean and uncertainty maps, and aggregate residual
diagnostics.

Run `01_probabilistic_disaggregation.ipynb` first to train the model
and save the posterior grid arrays.

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt


ROOT = Path("..").resolve() if Path.cwd().name == "notebooks" else Path(".").resolve()
DATA_PROCESSED = ROOT / "data" / "processed"
REPORTS_FIGURES = ROOT / "reports" / "figures"
REPORTS_FIGURES.mkdir(parents=True, exist_ok=True)
```

```python
mu_prior_grid = np.load(DATA_PROCESSED / "mu_prior_grid.npy")
std_prior_grid = np.load(DATA_PROCESSED / "std_prior_grid.npy")
mu_post_grid = np.load(DATA_PROCESSED / "mu_post_grid.npy")
std_post_grid = np.load(DATA_PROCESSED / "std_post_grid.npy")

print("Prior mean shape :", mu_prior_grid.shape)
print("Posterior mean shape:", mu_post_grid.shape)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

im0 = axes[0, 0].imshow(mu_prior_grid)
fig.colorbar(im0, ax=axes[0, 0])
axes[0, 0].set_title("Prior mean")

im1 = axes[0, 1].imshow(std_prior_grid)
fig.colorbar(im1, ax=axes[0, 1])
axes[0, 1].set_title("Prior std")

im2 = axes[1, 0].imshow(mu_post_grid)
fig.colorbar(im2, ax=axes[1, 0])
axes[1, 0].set_title("Posterior mean")

im3 = axes[1, 1].imshow(std_post_grid)
fig.colorbar(im3, ax=axes[1, 1])
axes[1, 1].set_title("Posterior std")

plt.tight_layout()
plt.savefig(REPORTS_FIGURES / "posterior_maps.png", dpi=150)
plt.show()
```

```python
# Uncertainty reduction: prior std - posterior std
uncertainty_reduction = std_prior_grid - std_post_grid

plt.figure(figsize=(6, 5))
plt.imshow(uncertainty_reduction, cmap="hot")
plt.colorbar(label="Prior std − Posterior std")
plt.title("Uncertainty reduction from coarse observations")
plt.tight_layout()
plt.savefig(REPORTS_FIGURES / "uncertainty_reduction.png", dpi=150)
plt.show()
```
